```python
# ============================================================
# Configure OneLake OAuth
# ============================================================

spark.conf.set(
    "fs.azure.account.auth.type",
    "OAuth"
)

spark.conf.set(
    "fs.azure.account.oauth.provider.type",
    "org.apache.hadoop.fs.azurebfs.oauth2.ClientCredsTokenProvider"
)

spark.conf.set(
    "fs.azure.account.oauth2.client.id",
    client_id
)

spark.conf.set(
    "fs.azure.account.oauth2.client.secret",
    client_secret
)

spark.conf.set(
    "fs.azure.account.oauth2.client.endpoint",
    (
        f"https://login.microsoftonline.com/"
        f"{tenant_id}/oauth2/token"
    )
)


print(
    "OneLake OAuth configuration applied."
)
```

```python
# ============================================================
# Test Fabric Gold Read
# ============================================================

FACT_SUPPLIER_PERFORMANCE_PATH = (
    "abfss://Project_Procurement@onelake.dfs.fabric.microsoft.com/lh_procurement_gold.Lakehouse/Tables/fact_supplier_performance"
)


fact_supplier_performance_df = (
    spark.read
    .format("delta")
    .load(
        FACT_SUPPLIER_PERFORMANCE_PATH
    )
)


print(
    "Rows read from Fabric Gold:",
    f"{fact_supplier_performance_df.count():,}"
)


display(
    fact_supplier_performance_df.limit(20)
)
```

```python
# ============================================================
# Validate Databricks -> Fabric OneLake Write Access
# ============================================================

from pyspark.sql import functions as F


# Derive the Gold Lakehouse Tables root from the working
# fact_supplier_performance ABFS path.
GOLD_TABLES_ROOT = (
    FACT_SUPPLIER_PERFORMANCE_PATH
    .rsplit("/Tables/", 1)[0]
    + "/Tables"
)


CONNECTIVITY_TEST_PATH = (
    f"{GOLD_TABLES_ROOT}/"
    f"monitoring_databricks_onelake_connectivity"
)


# Create one monitoring record
source_row_count = fact_supplier_performance_df.count()

connectivity_test_df = (
    spark.createDataFrame(
        [
            (
                "Azure Databricks",
                "Microsoft Fabric OneLake",
                "lh_procurement_gold",
                "fact_supplier_performance",
                source_row_count,
                True
            )
        ],
        [
            "SourcePlatform",
            "TargetPlatform",
            "TargetLakehouse",
            "ValidatedSourceTable",
            "ValidatedSourceRowCount",
            "ReadValidationPassed"
        ]
    )
    .withColumn(
        "ValidationTimestampUTC",
        F.current_timestamp()
    )
)


# Write the monitoring table directly into Fabric Gold
(
    connectivity_test_df
    .write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .save(CONNECTIVITY_TEST_PATH)
)


print(
    "Databricks -> Fabric OneLake write completed successfully."
)

print(
    "Target path:",
    CONNECTIVITY_TEST_PATH
)
```

```python
# ============================================================
# Validate Written Delta Table
# ============================================================

connectivity_validation_df = (
    spark.read
    .format("delta")
    .load(CONNECTIVITY_TEST_PATH)
)


validation_row_count = connectivity_validation_df.count()


if validation_row_count != 1:
    raise ValueError(
        f"Connectivity validation failed. "
        f"Expected 1 row but found {validation_row_count}."
    )


print(
    "Databricks -> Fabric write-back validation PASSED."
)

print(
    "Rows:",
    validation_row_count
)


display(
    connectivity_validation_df
)
```
